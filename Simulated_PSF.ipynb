# Construct the Radial Profile with simulated PSFs 

```python
# Import libraries 
import matplotlib, matplotlib.pyplot as plt
import stpsf
import pandas as pd 
%matplotlib inline
```

### (1) JWST/F115W

```python
# Instantiate a model of NIRCam filters 
nrc = stpsf.NIRCam()
nrc.filter =  'F277W'
psf = nrc.calc_psf(oversample=2,fov_pixels=67)     # returns an astropy.io.fits.HDUlist containing PSF and header
stpsf.display_psf(psf)   

```

```python
print(psf[1].header)
```

```python
plt.imshow(psf[0].data, norm=matplotlib.colors.LogNorm(1e-6, 1), origin = 'lower')  
```

```python
print(f"Extension has oversampling factor = ", psf[0].header['OVERSAMP'],
          f"\tPixelscale = {psf[0].header['PIXELSCL']:.4f} arcsec/pix",
          f"\tFWHM = {stpsf.measure_fwhm(psf):.4f} arcsec")
```

```python
# PSF sum for F115W 
psf = nrc.calc_psf(fov_arcsec=10)

print(f"PSF sum within {psf[0].header['FOV']:.1f} arcsec box for "
      f"{psf[0].header['FILTER']}: {psf[0].data.sum():.3f}")
```

```python
# Setup the plot area (optional, but good practice for side-by-side plots)
plt.figure(figsize=(10, 5))

# Plot the radial profile 
stpsf.display_ee(psf) 
plt.title(f"NIRCam {nrc.filter} Encircled Energy (Model)")
plt.ylabel("Encircled Energy (Normalized Flux)")
plt.xlim([-0.05,1.25])
plt.xlabel("Radius (arcsec)")

plt.tight_layout()
plt.show()
```

```python
# Save the results for selected filters as pandas dataframes
filter_list = ['f090w', 'f115w', 'f150w', 'f200w', 'f277w', 'f356w', 'f410m', 'f444w']
nrc = stpsf.NIRCam()
for i in range(len(filter_list)):
    nrc.filter =  filter_list[i]
    psf = nrc.calc_psf(oversample=4)    
    rr, ee = stpsf.measure_ee(psf) 
    df = pd.DataFrame({
    "radius_pixel": rr,
    "encircled_energy": ee
        })
    df.to_csv('/Users/quocanh1306/Downloads/Nircam_%s.csv' % filter_list[i])
```
